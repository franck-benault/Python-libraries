# NetworkX
Software for complex networks

```python
import networkx as nx



G = nx.Graph()
G.add_node(1)
G.add_nodes_from([2, 3, 4,5])

G.add_edge(1, 2, weight=5.0)
G.add_edge(1, 3)
G.add_edge(2, 4)
G.add_edge(3, 4)
G.add_edge(4, 4)
G.add_edge(4, 5)

print(G.nodes())            
# All edges
print(G.edges())     

```

```python
nx.draw_networkx(G)
```
